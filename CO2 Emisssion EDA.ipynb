# Univariate analysis

```python
# Select numerical columns
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns

# Determine the number of rows and columns for the subplots
n_cols = 4  # Number of histograms per row
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols  # Calculate required rows

# Create subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(24, n_rows * 6))  # Adjust figure size
axes = axes.flatten()

# Plot each numerical column
for i, col in enumerate(numerical_cols):
    ax = axes[i]
    df[col].hist(bins=20, ax=ax, edgecolor='black')
    ax.set_title(col, fontsize=14)  # Larger title font size
    ax.tick_params(axis='x', labelsize=12)  # Larger x-axis labels
    ax.tick_params(axis='y', labelsize=12)  # Larger y-axis labels
    ax.set_xlabel(col, fontsize=12)  # Add x-axis label
    ax.set_ylabel('Frequency', fontsize=12)  # Add y-axis label

# Remove unused subplots
for i in range(len(numerical_cols), len(axes)):
    fig.delaxes(axes[i])

# Adjust layout and add a main title
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.suptitle("Histograms of Numerical Features", fontsize=20, weight='bold')  # Larger main title
plt.show()
```

**1. General Observations**
The dataset focuses on CO2 emissions, energy consumption, and related metrics, both at absolute levels and normalized by population, GDP, or global share.Many features are highly skewed, dominated by low or zero values, indicating uneven distributions globally.

**2. Observations by Feature Categories**

These features show heavy right-skewness, with the majority of countries having lower emissions and a few countries contributing significantly more.
Outliers likely represent industrialized nations or large economies with high emissions.
Per Capita Emissions (coal_co2_per_capita, gas_co2_per_capita, oil_co2_per_capita, cement_co2_per_capita, co2_per_capita)
Similar to absolute emissions, most countries have low values with a long tail.
Developed nations or those with high fossil fuel dependency likely account for the tail-end outliers.
CO2 Growth Rates (co2_growth_abs, co2_growth_prct)
The distributions have peaks near zero, reflecting stable or slow-growing emissions in most countries.
Outliers represent rapid industrialization or sudden emission increases.

- **B. Cumulative CO2 Metrics :** Cumulative Emissions (cumulative_coal_co2, cumulative_gas_co2, cumulative_oil_co2, etc.)
These metrics exhibit highly skewed distributions, with long tails dominated by historically high emitters like industrialized nations.
Global Shares (share_global_coal_co2, share_global_gas_co2, etc.)

- **C. CO2 Emission Efficiency :** Emissions Per GDP (co2_per_gdp, co2_including_luc_per_gdp)
Distributions indicate that most countries have low emissions relative to GDP, but some countries (likely fossil fuel-dependent economies) have significantly higher ratios.
Emissions Per Unit Energy (co2_per_unit_energy, co2_including_luc_per_unit_energy)
These features are also skewed, reflecting differences in energy efficiency and reliance on fossil fuels.

- **D. Energy Consumption Features :** Primary Energy Consumption (primary_energy_consumption)
Heavily skewed distribution, with the majority of countries consuming small amounts of energy and a few consuming significantly more (likely industrialized nations).
Energy Per Capita and Per GDP (energy_per_capita, energy_per_gdp)
Similar patterns, reflecting disparities in energy access and efficiency across countries.


- **E. Trade and Consumption-based Metrics :** Trade CO2 Metrics (trade_co2, trade_co2_share)
Likely dominated by low values, with a small number of countries involved heavily in trade-related emissions.
Consumption-based CO2 (consumption_co2, consumption_co2_per_capita)
Skewed distributions reflect disparities in consumption patterns.


- **F. Temperature Change (temperature_change_from_co2) :** Most countries likely show small temperature changes relative to CO2 emissions, with outliers reflecting regions experiencing significant impacts.

- Population shows moderate to strong correlations with most emission variables

2. Weak or Negative Correlations (Blue):
- Trade-related variables (trade_co2, trade_co2_share) show weak or negative correlations with many other variables

- Growth-related variables (co2_growth_prct) show weak correlations with most other variables

3. Time Patterns:
- The 'year' variable shows notable correlations with many measures, suggesting significant temporal trends in emissions and related factors

4. Notable Relationships:
- Temperature change shows moderate positive correlations with cumulative emissions
- Energy consumption per capita correlates strongly with CO2 per capita
- Global share variables tend to correlate strongly with each other but weakly with per capita measures

5. Sector-Specific Patterns:
- Different emission sources (coal, gas, cement, etc.) show varying degrees of correlation with economic indicators
- Land use change emissions show somewhat different patterns from industrial emissions

This matrix reveals the complex interconnections between various aspects of CO2 emissions, energy use, and economic factors, suggesting that addressing climate change requires considering multiple interrelated factors.

```python
# A heatmap helps to visualize the strength of correlations between different variables.
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap between CO2 Emissions, GDP, Population, and Energy Consumption')
plt.show()
```

- 1. Strongest Relationships (Dark Red, close to 1.0):
  -     CO2 emissions and Energy Consumption show a very strong correlation (0.90)
  -     CO2 emissions and Population are also strongly correlated (0.88)
  -     Of course, each variable has a perfect correlation (1.00) with itself, shown in the diagonal red squares
- 2. Weakest Relationships (Blue, closer to 0.4) : GDP shows relatively weak correlations with all other variables:
  -     GDP vs CO2 emissions: 0.41
  -     GDP vs Population: 0.40
  -     GDP vs Energy Consumption: 0.47
- 3. Moderate to Strong Relationships : Population and Energy Consumption show a strong correlation (0.82)
- 4. Key Insights:
  -     Energy use and CO2 emissions are very tightly linked, suggesting that current energy production methods are still heavily carbon-dependent
  -     Population size has a strong influence on both emissions and energy consumption
  -     Surprisingly, GDP (economic output) has much weaker connections with emissions, energy use, and population, suggesting that economic                    development doesn't necessarily mean proportionally higher emissions or energy use

**This tells us that while population growth and energy use are big drivers of CO2 emissions, economic growth (GDP) might be able to happen without necessarily causing the same level of environmental impact.**
